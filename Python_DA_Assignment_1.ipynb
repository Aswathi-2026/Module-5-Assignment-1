```python

import numpy as np
import pandas as pd

print("\n========== PART 1: NUMPY ARRAY OPERATIONS ==========")
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("\nWeek 1 Temperatures:")
print(temperatures_w1)

print("\nShape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)

temperatures_fahrenheit = (temperatures_w1 * 9/5) + 32
print("\nTemperatures in Fahrenheit:")
print(temperatures_fahrenheit)

print("\nMaximum Temperature:", np.max(temperatures_w1))
print("Minimum Temperature:", np.min(temperatures_w1))
print("Mean Temperature:", np.mean(temperatures_w1))

print("\nTemperatures for the First Three Days:")
print(temperatures_w1[:3])

print("\nTemperatures for the Weekend (Last Two Days):")
print(temperatures_w1[-2:])

print("\nTemperatures for the Middle Three Days:")
print(temperatures_w1[2:5])

temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print("\nTwo-Week Temperature Array:")
print(temperatures)

print("\nShape of 2D Array:", temperatures.shape)
print("Data Type of 2D Array:", temperatures.dtype)
print("Total Number of Elements:", temperatures.size)

print("\nWeek 1 Temperatures:")
print(temperatures[0])

print("\nWeek 2 Temperatures:")
print(temperatures[1])

print("\nWeekend Temperatures for Week 1:")
print(temperatures[0, -2:])

print("\nWeekend Temperatures for Week 2:")
print(temperatures[1, -2:])

print("\n========== PART 2: PANDAS SERIES ==========")

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print("\nOriginal Marks Series:")
print(marks)

print("\nMark of the 1st Rank Student (Integer Position):")
print(marks.iloc[0])

print("\nMarks of the Top 3 Rank Students using loc:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

print("\nMark of the 3rd Rank Student using iloc:")
print(marks.iloc[2])

print("\nRanks with Marks Greater Than 90:")
print(marks[marks > 90])

marks.loc['Rank1'] = 100
print("\nMarks after Modifying Rank1 to 100:")
print(marks)

marks = marks.drop('Rank5')
print("\nMarks after Removing Rank5:")
print(marks)

cgpa = marks / 10
print("\nCGPA of Students:")
print(cgpa)

print("\n========== PART 3: PANDAS DATAFRAME ==========")

transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'
    ],
    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

print("\nOriginal Transactions DataFrame:")
print(transactions)

print("\nFirst 5 Rows (head):")
print(transactions.head())

print("\nLast 5 Rows (tail):")
print(transactions.tail())

print("\nShape of DataFrame:")
print(transactions.shape)

print("\nColumn Names:")
print(transactions.columns)

print("\nData Types:")
print(transactions.dtypes)

print("\nProductCategory and Amount Columns:")
print(transactions[['ProductCategory', 'Amount']])

print("\nLast 3 Columns using iloc:")
print(transactions.iloc[:, -3:])

print("\nRows where Region is North and Amount > 200:")
print(transactions[(transactions['Region'] == 'North') &
                   (transactions['Amount'] > 200)])

print("\nValue Counts of ProductCategory:")
print(transactions['ProductCategory'].value_counts())

print("\nUnique Values in Region:")
print(transactions['Region'].unique())

print("\nMean Amount Grouped by Region:")
print(transactions.groupby('Region')['Amount'].mean())

transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
print("\nAfter Modifying TransactionID 102 Amount to 165:")
print(transactions)

transactions['Discount'] = transactions['Amount'] * 0.10
print("\nAfter Adding Discount Column:")
print(transactions)

transactions = transactions[transactions['TransactionID'] != 109]
print("\nAfter Removing TransactionID 109:")
print(transactions)

transactions = transactions.drop(columns=['Discount'])
print("\nFinal Transactions DataFrame (Discount Column Deleted):")
print(transactions)

print("\n========== ASSIGNMENT COMPLETED SUCCESSFULLY! ==========")
```
